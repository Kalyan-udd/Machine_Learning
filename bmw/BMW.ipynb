```python
import pandas as pd
from sklearn.linear_model import LinearRegression, PoissonRegressor
from sklearn.ensemble import RandomForestRegressor,VotingRegressor,GradientBoostingRegressor
from sklearn.svm import SVR
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import r2_score, root_mean_squared_error
from sklearn.model_selection import train_test_split
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("bmw_cars_market_dataset_synthetic.csv")
df.head()
```

```python
df = df.dropna(axis=0)
df.head()
```

```python
df.columns
```

```python
df['model'].unique()
model_mapping = {}
k =0
for i in df['model']:
    model_mapping[i] = model_mapping.get(i,k)
    k+=1
df["model_id"] = df['model'].map(model_mapping)
```

```python
df['fuel_type'].unique()
fuel_map = {
    'diesel':0,
    'petrol':1,
    'electric':2,
    'hybrid':3
}

df['fuel_type_id'] = df['fuel_type'].map(fuel_map)
```

```python
df["transmission"].unique()
transmission_map = {
    'automatic':0,
    'manual':1
}
df["transmission_id"] = df["transmission"].map(transmission_map)
df["transmission_id"].head()
```

```python
df["drivetrain"].unique()
drivetrain_map = {
    'AWD':0, 
    'FWD':1, 
    'RWD':2
}
df["drivetrain_id"] = df["drivetrain"].map(drivetrain_map)
df["drivetrain_id"].tail()
```

```python
df["body_type"].unique()
body_map={
    'suv':0, 'hatchback':1, 'sedan':2, 'coupe':3, 'convertible':4
}
df["body_type_id"] = df["body_type"].map(body_map)
df['body_type_id'].tail()
```

```python
df['color'].unique()
color_map = {
    'white':0, 'black':1, 'blue':2, 'silver':3, 'grey':4, 'red':5
}
df['color_id'] = df['color'].map(color_map)
df['color_id'].tail()
```

```python
accident_history_map = {
    'no':0, 'yes':1
}
df['accident_history_id'] = df['accident_history'].map(accident_history_map)
df['accident_history_id'].unique()
```

```python
service_history_map = {
    'partial':0, 'full':1, 'none':2
}
df['service_history_id'] = df['service_history'].map(service_history_map)
df['service_history_id'].unique()
```

```python
country_sold_map = {}
m = 0
for i in df['country_sold']:
    if i not in country_sold_map:
        country_sold_map[i] = country_sold_map.get(i,m)
        m+=1
df["country_sold_id"] = df['country_sold'].map(country_sold_map)
df['country_sold_id'].unique()
```

```python
y = df.price_usd
x = df[['car_id', 'model_id', 'year', 'engine_size', 'horsepower', 'fuel_type_id',
       'transmission_id', 'drivetrain_id', 'mileage_km',
       'fuel_consumption_l_per_100km', 'co2_emissions_g_km',
       'doors', 'seats', 'body_type_id', 'color_id', 'owner_count',
       'accident_history_id', 'service_history_id', 'country_sold_id']]
```

Data splitting for fitting models...

```python
df_IDs = df[['car_id', 'model_id', 'year', 'engine_size', 'horsepower', 'fuel_type_id',
       'transmission_id', 'drivetrain_id', 'mileage_km',
       'fuel_consumption_l_per_100km', 'co2_emissions_g_km',
       'doors', 'seats', 'body_type_id', 'color_id', 'owner_count',
       'accident_history_id', 'service_history_id', 'country_sold_id', 'price_usd']]

df_corr = df_IDs.corr(method='kendall')
plt.figure(figsize=(15,12))
sns.heatmap(df_corr,cmap='magma')
plt.legend()
plt.show()
```

Every parameter has some effect on the price hence for the accuracy we need to use all the parameters changed to numbers.

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
linear_model = LinearRegression()
linear_model.fit(x_train,y_train)

y_pred_linear = linear_model.predict(x_test)

R2_linear = r2_score(y_test,y_pred_linear)
rmse_linear = root_mean_squared_error(y_test,y_pred_linear)

R2_linear,rmse_linear
```

```python
plt.figure(figsize=(8,6))
sns.lineplot(data=x_test, x='model_id', y=y_test, color='blue',label='Actual Value',errorbar=None)
sns.lineplot(data=x_test, x='model_id', y=y_pred_linear,color='red', linestyle='--',label='predicted value', errorbar=None)

plt.show()
```

```python
poisson_model = PoissonRegressor()
poisson_model.fit(x_train,y_train)
y_pred_poisson = poisson_model.predict(x_test)

R2_poisson = r2_score(y_test, y_pred_poisson)
rmse_poisson = root_mean_squared_error(y_test,y_pred_poisson)

R2_poisson, rmse_poisson
```

```python
sns.lineplot(data=x_test,x="model_id",y=y_test,color='blue',errorbar=None)
sns.lineplot(data=x_test,x='model_id',y=y_pred_poisson,color='red', linestyle='--',errorbar=None)
```

```python
rf_model = RandomForestRegressor()
rf_model.fit(x_train,y_train)
y_predicted_rf = rf_model.predict(x_test)

r2_rf = r2_score(y_test,y_predicted_rf)
r2_rf
```

```python
sns.lineplot(data=x_test,x='model_id',y=y_test,color='blue',errorbar=None)
sns.lineplot(data=x_test,x='model_id',y=y_predicted_rf,color='red',linestyle='--',errorbar=None)
```

```python
dtree_model = DecisionTreeRegressor()
dtree_model.fit(x_train,y_train)
y_pred_dtree = dtree_model.predict(x_test)

r2_dtree = r2_score(y_test,y_pred_dtree)
r2_dtree
```

```python
sns.lineplot(data=x_test,x='model_id',y=y_test,color='blue',errorbar=None)
sns.lineplot(data=x_test,x='model_id',y=y_pred_dtree,color='red',linestyle='--',errorbar=None)
```

```python
scaler = StandardScaler()
scaled_x_train = scaler.fit_transform(x_train)
scaled_x_test = scaler.transform(x_test)

svr_model = SVR(kernel='linear',C=10,epsilon=0.0001)
svr_model.fit(scaled_x_train,y_train)
y_pred_svr = svr_model.predict(scaled_x_test)

r2_svr = r2_score(y_test,y_pred_svr)
r2_svr
```

```python
sns.lineplot(data=x_test,x='model_id',y=y_test,color='blue',errorbar=None)
sns.lineplot(data=x_test,x='model_id',y=y_pred_svr,color='red',linestyle='--',errorbar=None)
```

```python
GBR_model = GradientBoostingRegressor()
GBR_model.fit(x_train,y_train)
y_pred_gbr = GBR_model.predict(x_test)

r2_gbr = r2_score(y_test,y_pred_gbr)
r2_gbr
```

```python
sns.lineplot(data=x_test,x='model_id',y=y_test,color='blue',errorbar=None)
sns.lineplot(data=x_test,x='model_id',y=y_pred_gbr,color='red',linestyle='--',errorbar=None)
```
