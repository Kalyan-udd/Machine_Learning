```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, root_mean_squared_error
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("biofuel_nanoblend_dataset.csv")
df.head()
```

```python
df.describe(include=["object", "float"])
```

```python
df.columns
```

```python
df["nanoparticle_type"].describe()
```

Here we have some objects in the dataset and since mathematical models are only capable of understanding numbers, we make another column to make a proxy of it by mapping some numbers to the objects.


```python
nano_map = {
    "TiO2" : 1,
    "ZnO" : 2,
    "CeO2" : 3
}


df["nano_particle_id"] = df["nanoparticle_type"].map(nano_map)
```

```python
df.describe()
```

```python
df.columns
```

```python
df.dropna(axis=0)
```

```python
x = df[['engine_load', 
        'fuel_blend_percentage',
        'nanoparticle_concentration', 
        'injection_pressure', 
        'engine_speed', 
        'CO_emissions', 
        'NOx_emissions', 
        'HC_emissions', 
        'CO2_emissions',
        'smoke_opacity', 
        'nano_particle_id']]
y = df['BTE']

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

LINEAR REGRESSION MODEL

```python
Biofuel_model = LinearRegression()
Biofuel_model.fit(x_train,y_train)
y_predicted = Biofuel_model.predict(x_test)
R2_score = r2_score(y_test, y_pred=y_predicted)
Rmse = root_mean_squared_error(y_test, y_predicted)

R2_score, Rmse
```

DECISION TREE MODEL

```python
Biofuel_model_decisiontree = DecisionTreeRegressor()
Biofuel_model_decisiontree.fit(x_train,y_train)
y_predicted_tree = Biofuel_model_decisiontree.predict(x_test)

R2_tree = r2_score(y_test, y_pred=y_predicted_tree)
rmse_tree = root_mean_squared_error(y_test, y_predicted_tree)

R2_tree, rmse_tree
```

RANDOM FOREST REGRESSOR


```python
Biofuel_model_forest = RandomForestRegressor()
Biofuel_model_forest.fit(x_train, y_train)

y_predicted_forest = Biofuel_model_forest.predict(x_test)

R2_score_forest = r2_score(y_test, y_predicted_forest)
rmse_forest = root_mean_squared_error(y_test, y_predicted_forest)

R2_score_forest, rmse_forest
```
