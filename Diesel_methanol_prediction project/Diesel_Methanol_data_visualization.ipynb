```python

import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
DataFrame = pd.read_csv("Diesel_Methanol_clean_data.csv")
DataFrame
```

```python
DataFrame = DataFrame.drop(columns=["Unnamed: 0"])
DataFrame
```

```python
DataFrame.plot(kind='bar', x='Fuel', y='Brake_Thermal_Efficiency_(BTE)_in_%', color='teal', legend=False)

plt.title("Relationship Between Fuel type and Break Thermal Efficiency")
plt.xlabel("Fuel Type")
plt.ylabel("Brake Thermal Efficiency")

plt.grid(axis='y')

plt.show()
```

```python
sns.regplot(x="Brake_Power_(KW)", y='Brake_Thermal_Efficiency_(BTE)_in_%', data=DataFrame, order=2)
plt.title("Relation Between Breaking power and Braking Thermal Efficiency")
plt.xlabel("Break power in KW")
plt.ylabel("Break Thermal Efficiency in %")
plt.show()
```

```python
sns.regplot(x="Injection_Pressure_(IP)_in_Bar", y='Brake_Thermal_Efficiency_(BTE)_in_%', data=DataFrame, order=2)
plt.title("Relation Between Injection pressure and Braking Thermal Efficiency")
plt.xlabel("Injection pressure in bar")
plt.ylabel("Break Thermal Efficiency in %")
plt.show()
```

```python
DataFrame.plot(kind='bar' ,x='Fuel', y='Unburnt_Hydrocarbons_(UHC)_in_PPM', color='teal', legend=False)
plt.title("Relation between Fuel type and the Unburnt hydrocarbon concentraion in PPM")
plt.xlabel("Fuel")
plt.ylabel("Unburnt hydrocarbons in PPM")

plt.show()
```

```python
sns.regplot(x="Injection_Pressure_(IP)_in_Bar", y='Unburnt_Hydrocarbons_(UHC)_in_PPM', data=DataFrame, order=2)
plt.title("Relation Between Injection pressure and Unburnt hydrocarbons")
plt.xlabel("Injection pressure in bar")
plt.ylabel("Unburnt hydrocarbons in ppm")
plt.show()
```

```python
sns.regplot(x="Brake_Power_(KW)", y='Unburnt_Hydrocarbons_(UHC)_in_PPM', data=DataFrame, order=2)
plt.title("Relation Between Breaking power and Unburnt hydrocarbons")
plt.xlabel("Break power in KW")
plt.ylabel("Unburnt hydrocarbons in ppm")
plt.show()
```

```python
DataFrame.plot(kind='bar' ,x='Fuel', y='Smoke_in_HSU', color='teal', legend=False)
plt.title("Relation between Fuel type and the Smoke in HSU")
plt.xlabel("Fuel")
plt.ylabel("Smoke in HSU")

plt.show()
```

```python
sns.regplot(x="Brake_Power_(KW)", y='Smoke_in_HSU', data=DataFrame, order=2)
plt.title("Relation Between Breaking power and Smoke in HSU")
plt.xlabel("Break power in KW")
plt.ylabel("Smoke in HSU")
plt.show()
```

```python
sns.regplot(x="Injection_Pressure_(IP)_in_Bar", y='Smoke_in_HSU', data=DataFrame, order=2)
plt.title("Relation Between injection pressure and Smoke in HSU")
plt.xlabel("Injection pressure in bar")
plt.ylabel("Smoke in HSU")
plt.show()
```

```python
DataFrame.plot(kind='bar' ,x='Fuel', y='Nitrogen_Oxides_(NOx)_in_g/kWh', color='teal', legend=False)
plt.title("Relation between Fuel type and the Nitrogen oxides in g/kwh")
plt.xlabel("Fuel")
plt.ylabel("Nitrogen oxides in g/kwh")

plt.show()
```

```python
DataFrame.plot(kind='bar' ,x='Fuel_id', y='Nitrogen_Oxides_(NOx)_in_g/kWh', color='teal', legend=False)
plt.title("Relation between Fuel type and the Nitrogen oxides in g/kwh")
plt.xlabel("Fuel")
plt.ylabel("Nitrogen oxides in g/kwh")

plt.show()
```

```python
sns.regplot(x="Injection_Pressure_(IP)_in_Bar", y='Nitrogen_Oxides_(NOx)_in_g/kWh', data=DataFrame, order=2)
plt.title("Relation Between Breaking power and Nitrogen oxides in g/kWh")
plt.xlabel("Injection pressure in bar")
plt.ylabel("Nitrogen oxides in g/kWh")
plt.show()
```

```python
# 3. Visualizing NOx emissions scaling with Load
plt.figure(figsize=(10, 6))
sns.lineplot(data=DataFrame, x='Brake_Power_(KW)', y='Nitrogen_Oxides_(NOx)_in_g/kWh', hue='Fuel', marker='o', linewidth=2.5)

# Adding the 'Uncertainty' shaded region like the paper (±0.92%)
# Based on the paper's TSU for NOx [cite: 331, 440]
plt.title("NOx Emission Trends Across Load Conditions")
plt.ylabel("NOx (g/kWh)")
plt.xlabel("Engine Load (%)")
plt.show()
```
