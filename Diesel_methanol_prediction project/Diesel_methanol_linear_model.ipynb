```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split 
from sklearn.metrics import r2_score, root_mean_squared_error
import seaborn as sns
import matplotlib.pyplot as plt

```

```python
df = pd.read_csv("Diesel_Methanol_clean_data.csv")

df = df.drop(columns=['Unnamed: 0'])
df
```

```python
df.columns
```

```python
x = df[['Brake_Power_(KW)', 'Injection_Pressure_(IP)_in_Bar','Fuel_id']]
y_BTE = df['Brake_Thermal_Efficiency_(BTE)_in_%']

x_test,x_train,y_test_BTE,y_train_BTE = train_test_split(x,y_BTE,test_size=0.2,random_state=42)
```

```python
linear_model = LinearRegression()
linear_model.fit(x_train,y_train_BTE)
y_pred_BTE = linear_model.predict(x_test)

R2_score = r2_score(y_test_BTE,y_pred_BTE)
RMSE = root_mean_squared_error(y_test_BTE, y_pred_BTE)

print(df['Brake_Thermal_Efficiency_(BTE)_in_%'].std())
R2_score, RMSE
```

Hence the linear regression model is very dumb since we have root mean square error more than standard deviation

```python
plt.figure(figsize=(10,6))
sns.lineplot(data=x_test,x='Brake_Power_(KW)', y=y_test_BTE,markers='o', color='blue',label='Actual Values',errorbar=None)
sns.lineplot(data=x_test, x='Brake_Power_(KW)', y=y_pred_BTE, markers='x',color='red',linestyle='--', label='Predicted Values',errorbar=None)
plt.title("Deviation Analysis: Actual vs. Predicted")
plt.grid(True, alpha=0.3)
plt.legend()

plt.show()
```
